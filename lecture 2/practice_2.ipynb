```python
import random
```

```python
''' Реализовать функцию (или набор функций),
которые выводят на экран таблицы умножения, деления, вычитания, сложения для чисел 1..9.
Выбор таблицы осуществляется пользователем '''
```

```python
def tab(n, znak):
    for i in range(1, 10):
        if znak == "+":
            print(n, "+", i, "=", n + i)
        if znak == "-":
            print(n, "-", i, "=", n - i)
        if znak == "*":
            print(n, "*", i, "=", n * i)
        if znak == "/":
            print(n, "/", i, "=", round(n / i, 2))
```

```python
print("1 - +")
print("2 - -")
print("3 - *")
print("4 - /")
v = int(input("Выбери: "))
n = int(input("Число: "))
if v == 1:
    tab(n, "+")
elif v == 2:
    tab(n, "-")
elif v == 3:
    tab(n, "*")
elif v == 4:
    tab(n, "/")
else:
    print("Ошибка")
```

```python
''' a) Создающая вектор заданной длины N, заполнение – случайные числа от 0..1 '''
```

```python
def a(n):
    x = []
    for i in range(n):
        x.append(random.random())
    return x
```

```python
n = int(input("Введите N: "))
print(a(n))
```

```python
'''  b) Создающая матрицу MxN, заполнение – случайные числа 0..1 '''
```

```python
def a(m, n):
    x = []
    for i in range(m):
        y = []
        for j in range(n):
            y.append(random.random())
        x.append(y)
    return x
```

```python
m = int(input("M = "))
n = int(input("N = "))
print(a(m, n))
```

```python
'''  c) Умножающая матрицу на вектор (см. предыдущее ДЗ) '''
```

```python
def a(mat, vector):
    result = []
    for i in range(len(mat)):
        s = 0
        for j in range(len(vector)):
            s += mat[i][j] * vector[j]
        result.append(s)
    return result
```

```python
n = int(input("введите число строк: "))
m = int(input("введите число столбцов: "))
mat = []
print("введите матрицу:")
for i in range(n):
    mat.append(list(map(float, input().split())))
print("введите вектор:")
vector = list(map(float, input().split()))
print("результат:")
print(a(mat, vector))
```

```python
'''  f) Находящую сумму диагональных элементов матрицы '''
```

```python
def a(x):
    s = 0
    for i in range(min(len(x), len(x[0]))):
        s += x[i][i]
    return s
```

```python
m = int(input("Строки: "))
n = int(input("Столбцы: "))
x = []
for i in range(m):
    y = list(map(int, input().split()))
    x.append(y)
print("Сумма =", a(x))
```

```python
'''  g) Реализующая двумерную свертку изображения.'''
```

```python
def a(img, ker):
    n = len(img)
    m = len(img[0])
    k = len(ker)
    result = []
    for i in range(n - k + 1):
        row = []
        for j in range(m - k + 1):
            s = 0
            for x in range(k):
                for y in range(k):
                    s += img[i + x][j + y] * ker[x][y]
            row.append(s)
        result.append(row)
    return result
```

```python
n = int(input("число строк изображения: "))
m = int(input("число столбцов изображения: "))
img = []
print("введите изображение:")
for i in range(n):
    img.append(list(map(float, input().split())))
k = int(input("размер ядра: "))
ker = []
print("введите ядро:")
for i in range(k):
    ker.append(list(map(float, input().split())))
print("результат:")
print(a(img, ker))
```
