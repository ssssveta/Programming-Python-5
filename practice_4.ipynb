```python
import random
```

```python
class Player:
    def __init__(self,name):
        self.name=name

    def guess(self,a,b):
        return 0
```

```python
class RandomPlayer(Player):
    def guess(self,a,b):
        return random.randint(a,b)
```

```python
class DichPlayer(Player):
    def guess(self,a,b):
        return (a+b)//2
```

```python
class UserPlayer(Player):
    def guess(self,a,b):
        return int(input(f"{self.name}, введи число от {a} до {b}: "))
```

```python
def game(p,n,a,b):
    k=0
    while a<=b:
        x=p.guess(a,b)
        k+=1
        if x==n:
            print(p.name,"угадал за",k,"попыток")
            return
        if x<n:
            a=x+1
        else:
            b=x-1
    print("Число не найдено")
```

```python
a=0
b=100
n=random.randint(a,b)
print("Загадано число от",a,"до",b)

ps=[RandomPlayer("Случайный"),DichPlayer("Дихотомия"),UserPlayer("Пользователь")]
for p in ps:
    print("\n",p.name)
    game(p,n,a,b)
```

```python
class Igra:
    def __init__(self, name):
        self.name = name

    def hod(self, a, b):
        return 0

    def otvet(self, x, otv):
        pass
```

```python
class Dich(Igra):
    def __init__(self, name, a, b):
        super().__init__(name)
        self.a = a
        self.b = b

    def hod(self, a, b):
        return (self.a + self.b) // 2

    def otvet(self, x, otv):
        if otv == "больше":
            self.a = x + 1
        elif otv == "меньше":
            self.b = x - 1
```

```python
class User(Igra):
    def hod(self, a, b):
        return int(input(f"{self.name}, введи число от {a} до {b}: "))
```

```python
class Manager:
    def __init__(self, a, b, k):
        self.a = a
        self.b = b
        self.k = k
        self.ps = []

    def add(self, p):
        self.ps.append(p)

    def game(self):
        bot = self.ps[0]
        user = self.ps[1]

        n1 = random.randint(self.a, self.b)
        n2 = int(input(f"Загадай число от {self.a} до {self.b}: "))

        print("\nИгра началась")

        a = self.a
        b = self.b

        for i in range(1, self.k + 1):
            print("\nШаг", i)

            x = bot.hod(a, b)
            print("Бот называет:", x)

            if x == n2:
                print("Бот угадал число")
                print("Победил: Бот")
                return

            otv = input("Для бота: больше или меньше? ")
            bot.otvet(x, otv)

            x = user.hod(a, b)
            print("Пользователь называет:", x)

            if x == n1:
                print("Пользователь угадал число")
                print("Победил: Пользователь")
                return

            if x < n1:
                print("Для пользователя: больше")
            else:
                print("Для пользователя: меньше")

        print("\nИгра остановлена")
        print("Достигнуто максимальное количество шагов:", self.k)
```

```python
a = int(input("Введите начало диапазона: "))
b = int(input("Введите конец диапазона: "))
k = int(input("Введите количество шагов: "))

bot = Dich("Бот", a, b)
user = User("Пользователь")

m = Manager(a, b, k)
m.add(bot)
m.add(user)
m.game()
```
