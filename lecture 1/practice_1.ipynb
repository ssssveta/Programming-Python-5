```python
low = int(input("нижняя граница: "))
high = int(input("верхняя граница: "))
while low <= high:
    mid = (low + high) // 2
    print(f"число равно {mid}?")
    ans = input("равно / меньше / больше: ").strip().lower()
    if ans == "равно":
        print(f"загаданное число: {mid}")
        break
    elif ans == "меньше":
        high = mid - 1
    elif ans == "больше":
        low = mid + 1
```

```python
d = [random.randint(1, 100) for i in range(100)]
print("выборка:", d)
mini = min(d)
maxi = max(d)
s_d= sorted(d)
n = len(s_d)
if n % 2 == 1:
    med = s_d[n // 2]
else:
    med = (s_d[n // 2 - 1] + s_d[n // 2]) / 2
print("минимум:", mini)
print("максимум:", maxi)
print("медиана:", med)
```

```python
n = int(input("размерность: "))
s = float(input("скаляр: "))
a = [random.randint(1, 10) for i in range(n)]
b = [random.randint(1, 10) for i in range(n)]
print("вектор a:", a)
print("вектор b:", b)
summ = [a[i] + b[i] for i in range(n)]
mult = [a[i] * b[i] for i in range(n)]
print("сумма:", summ)
print("произведение:", mult)
na = math.sqrt(sum(x * x for x in a))
nb = math.sqrt(sum(x * x for x in b))
print("норма a:", round(na, 3))
print("норма b:", round(nb, 3))
v = a if na > nb else b
print("вектор с большей нормой:", v)
print("умножение на скаляр:", [x * s for x in v])
```

```python
n = int(input("введите число строк матрицы: "))
m = int(input("введите число столбцов матрицы: "))
print(f"введите элементы матрицы {n}x{m}:")
mat = []
for i in range(n):
    mat.append(list(map(float, input(f"строка {i+1} ({m} чисел через пробел): ").split())))
print(f"введите элементы вектора размерности {m}:")
vector = list(map(float, input(f"{m} чисел через пробел: ").split()))
result = []
for i in range(n):
    s = 0
    for j in range(m):
        s += mat[i][j] * vector[j]
    result.append(s)
print("матрица:")
for i in mat:
    print(i)
print("вектор:")
print(vector)
print("результат умножения матрицы на вектор:")
print(result)
```

```python
a = list(map(float, input("введите элементы списка через пробел: ").split()))
pos = []
for i in range(len(a)):
    if a[i] > 0:
        pos.append(i)
res = a[:]
if pos:
    for i in range(len(a)):
      if a[i] < 0:
            left = [p for p in pos if p < i]
            right = [p for p in pos if p > i]
            if left and right:
                res[i] = (a[left[-1]] + a[right[0]]) / 2
            elif left:
                res[i] = a[left[-1]]
            elif right:
                res[i] = a[right[0]]
else:
    print("в списке нет положительных чисел ")
print("результат:", res)
```

```python
A = list(map(float, input("введите массив A: ").split()))
F = list(map(float, input("введите ядро F: ").split()))
n = len(A)
m = len(F)
half = m // 2
B = []
for i in range(n):
    s = 0
    for j in range(m):
        idx = i + j - half
        if 0 <= idx < n:
            s += A[idx] * F[j]
    B.append(s)
print("результат свертки B:", B)
```
